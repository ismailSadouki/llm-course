```python
# 50 tokens of each vocabulary, picked at regular intervals; and a quick look at the first ids
for name, t in pretrained.items():
    v = t.get_vocab()
    by_id = sorted(v.items(), key=lambda kv: kv[1])
    step = max(1, len(by_id) // 50)
    print(f"{name:22} {len(v):6} entries | first by id: {[x for x, _ in by_id[:6]]}")
    print("   sample:", [x for x, _ in by_id[::step][:50]])
```

```python
import re
def kind(token):
    t = token.lstrip("Ġ▁").removeprefix("##")
    return "word-like" if re.fullmatch(r"[A-Za-z]{3,}", t) else "other"
pd.DataFrame({name: pd.Series([kind(x) for x in t.get_vocab()]).value_counts(normalize=True) for name, t in pretrained.items()}).round(3)
```
